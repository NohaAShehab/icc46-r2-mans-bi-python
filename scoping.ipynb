## 🌐 Global Variables

### Using `global` Keyword

To modify a global variable from inside a function, you must use the `global` keyword. Without it, Python creates a new local variable instead.


```python
def modify_track():
    global track
    track = "Power bi "

modify_track()
print(track)
```

---

## 🏠 Local Variables

### Accessing Outer Function Variables

Local variables in an outer function can be **read** by inner functions, but cannot be modified without special keywords.


# local

### The `nonlocal` Keyword

To modify a variable from an outer (but not global) function scope, use the `nonlocal` keyword. This tells Python to use the variable from the enclosing scope rather than creating a new local variable.


```python
def outer():
    course = "Python"
    def print_course():
        print(f"from outer course : {course}")


outer()
```

## 🔗 Deeply Nested Functions

### `global` vs `nonlocal` in Nested Functions

- `global` - Accesses/modifies variables at the module level
- `nonlocal` - Accesses/modifies variables from the nearest enclosing (non-global) scope

These keywords work even in deeply nested function hierarchies.


```python
def outer():
    """ local variable can be accessed anywhere inside the function even if the inner functions """
    course = "Python" # local variable
    def print_course():
        print(f"from outer course : {course}")

    print_course()

outer()
```

```python
def outer():
    """ local variable can be accessed anywhere inside the function even if the inner functions """
    course = "Python" # local variable
    def modify_course():
        course = "intro to python" # new local varaible associated with the modify function
        print(f"course = {course}")

    modify_course()
    print(f"---- after modification course= {course}")

outer()
```

## 📝 Summary

**Variable Scope Hierarchy:**
1. **Local scope** - Variables inside a function
2. **Enclosing scope** - Variables in outer functions (nonlocal)
3. **Global scope** - Variables at module level (global)
4. **Built-in scope** - Python's built-in names

**Keywords:**
- `global` - Modify global variables from any scope
- `nonlocal` - Modify variables from the nearest enclosing (non-global) scope

**Key Rules:**
- ✅ Inner functions can **read** outer function variables
- ✅ Use `global` to modify global variables
- ✅ Use `nonlocal` to modify outer function variables
- ✅ Without these keywords, assignment creates a new local variable


```python
"I need to modify the local variable from inside the inner function"
def outer():
    """ local variable can be accessed anywhere inside the function even if the inner functions """
    course = "Python" # local variable
    def modify_course():
        nonlocal course  #please don't create new local variable just use the outer function's one
        course = "intro to python"
        print(f"course = {course}")

    modify_course()
    print(f"---- after modification course= {course}")

outer()
```
