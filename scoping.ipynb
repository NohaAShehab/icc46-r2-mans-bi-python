```python
def outer():
    """ local variable can be accessed anywhere inside the function even if the inner functions """
    course = "Python" # local variable
    def print_course():
        print(f"from outer course : {course}")

    print_course()

outer()
```

```python
def outer():
    """ local variable can be accessed anywhere inside the function even if the inner functions """
    course = "Python" # local variable
    def modify_course():
        course = "intro to python" # new local varaible associated with the modify function
        print(f"course = {course}")

    modify_course()
    print(f"---- after modification course= {course}")

outer()
```

```python
"I need to modify the local variable from inside the inner function"
def outer():
    """ local variable can be accessed anywhere inside the function even if the inner functions """
    course = "Python" # local variable
    def modify_course():
        nonlocal course  #please don't create new local variable just use the outer function's one
        course = "intro to python"
        print(f"course = {course}")

    modify_course()
    print(f"---- after modification course= {course}")

outer()
```
