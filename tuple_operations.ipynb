# ⚠️ Creating Single-Element Tuples

## Common Mistake

Without a comma, Python treats parentheses as grouping, not a tuple:

```python
myt = ("new")  # This is a string, not a tuple!
```


````python
myt = ("new",)
print(myt)
````

## ✅ Correct Way

Add a **comma** to create a single-element tuple:

```python
myt = ("new",)  # This is a tuple!
```


````python
newtt = tuple("new")
print(newtt)
````

## 🔤 Convert String to Tuple

Using `tuple()` converts a string into a tuple of individual characters:


````python
newtt = tuple(["new"])
print(newtt)
````
