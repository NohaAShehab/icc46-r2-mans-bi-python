First, let's see how we might access class data from an instance method:

Class methods can return values, just like regular methods:


```python

class Application:
    count =0  # class variable, represent property related to the class
    def __init__(self, name, email):
        self.name = name
        self.email = email
        Application.count +=1  # create instance variable for the object

print(Application.count)
```

### Class Methods as Factory Methods

Class methods are often used as factory methods to create instances in different ways:


```python
class Application:
    count =0  # class variable, represent property related to the class
    def __init__(self, name, email):
        self.name = name
        self.email = email
        print(self.__class__)
        self.__class__.count +=1  # create instance variable for the object

print(Application.count)
app4 = Application('e', 'we')
```

### Class Method to Access Class Data

We can create a class method to retrieve all stored instances:


```python
class Car:
    count  =0
    def __init__(self, model, color):
        self.model = model
        self.color = color
        self.__class__.count += 1

    # function return with total number of cars
    def get_number_of_cars(self):
        return  self.__class__.count


c= Car("we", "dsd")
c.get_number_of_cars()
```

```python
class Car:
    count  =0
    def __init__(self, model, color):
        self.model = model
        self.color = color
        self.__class__.count += 1

    # function return with total number of cars
    """
        decorator --> object from special function ,,
        change default behaviour of function..
    """
    """ you call the function using class """
    @classmethod
    def get_number_of_cars(cls):
        """ classmethod decorator effect
        first parameter of the function --> represent current class
        not the instance
        """
        print(cls)
        # return Car.count
        return cls.count


Car.get_number_of_cars()
```

```python
"""
 class methods can be used as object factory
"""
class Car:
    count  =0
    def __init__(self, model, color):
        self.model = model
        self.color = color
        self.__class__.count += 1

    @classmethod
    def create_default_car(cls):
        #return Car("tret", "wrr")
        return  cls("BMW", "Blue")

    @classmethod
    def create_red_car(cls, model):
        return cls(model, "Red")

cc = Car.create_red_car("KIA")

cc= Car("tret", "wrr")
```
