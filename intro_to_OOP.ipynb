### <span style='color:#F57C00'>🏭 Class Methods as Factory Methods</span>

<div style='background-color:#FFF3E0; padding:12px; border-radius:8px; border-left:4px solid #FF6F00'>
<p style='color:#E65100'><span style='color:#FF6F00'><strong>Class methods</strong></span> are often used as <span style='color:#F57C00'><strong>factory methods</strong></span> to create instances in <span style='color:#FF8F00'><strong>different ways</strong></span>:</p>

```python
class Application:
    count =0  # class variable, represent property related to the class
    def __init__(self, name, email):
        self.name = name
        self.email = email
        print(self.__class__)
        self.__class__.count +=1  # create instance variable for the object

print(Application.count)
app4 = Application('e', 'we')
```

### <span style='color:#C2185B'>🔍 Class Method to Access Class Data</span>

<div style='background-color:#FCE4EC; padding:10px; border-radius:6px'>
<p style='color:#AD1457'>We can create a <span style='color:#C2185B'><strong>class method</strong></span> to <span style='color:#E91E63'><strong>retrieve all stored instances</strong></span>:</p>
</div>


```python
class Car:
    count  =0
    def __init__(self, model, color):
        self.model = model
        self.color = color
        self.__class__.count += 1

    # function return with total number of cars
    def get_number_of_cars(self):
        return  self.__class__.count


c= Car("we", "dsd")
c.get_number_of_cars()
```

```python
""" create function depend on the class only no need to create instance so I can call the function """
class Car:
    count  =0
    def __init__(self, model, color):
        self.model = model
        self.color = color
        self.__class__.count += 1

    # function return with total number of cars
    """
        decorator --> object from special function ,,
        change default behaviour of function..
    """
    """ you call the function using class """
    @classmethod
    def get_number_of_cars(self):
        print(self)


Car.get_number_of_cars()
```

```python
class Car:
    count  =0
    def __init__(self, model, color):
        self.model = model
        self.color = color
        self.__class__.count += 1

    # function return with total number of cars
    """
        decorator --> object from special function ,,
        change default behaviour of function..
    """
    """ you call the function using class """
    @classmethod
    def get_number_of_cars(cls):
        """ classmethod decorator effect
        first parameter of the function --> represent current class
        not the instance
        """
        print(cls)
        # return Car.count
        return cls.count


Car.get_number_of_cars()
```

```python
"""
 class methods can be used as object factory
"""
class Car:
    count  =0
    def __init__(self, model, color):
        self.model = model
        self.color = color
        self.__class__.count += 1

    @classmethod
    def create_default_car(cls):
        #return Car("tret", "wrr")
        return  cls("BMW", "Blue")

    @classmethod
    def create_red_car(cls, model):
        return cls(model, "Red")

cc = Car.create_red_car("KIA")

cc= Car("tret", "wrr")
```
