## 📖 Reading Files

### Reading Entire File Content

The `read()` method reads the entire content of a file as a single string. Note that after reading, the file pointer moves to the end of the file.


```python
""" you need to define , r --> read, w--> write, a--> append """
def open_file(file_name, mode='r'):
    # I need to open file for reading ?
    try:
        file_object = open(file_name, mode)
        print(file_object)
        # read content  into one string
        data = file_object.read() # data ??
        print(data, type(data))
        # close the file after reading
        file_object.close()
    except Exception as e:
        print(e)
        print("--- make sure that file exists ---")

open_file("users.txt", "r")
```

### File Pointer Position

When you read from a file, the file pointer moves forward. To read the file again, you need to reset the pointer using `seek(0)`.


```python
""" you need to define , r --> read, w--> write, a--> append """
def open_file(file_name, mode='r'):
    # I need to open file for reading ?
    try:
        file_object = open(file_name, mode)
        print(file_object)
        # read content  into one string
        data = file_object.read() # data ??
        print(data, type(data))
        print("*****************************")
        data2 = file_object.read()
        print(data2, type(data2), len(data2))
        file_object.close()
        # close the file after reading
        file_object.close()
    except Exception as e:
        print(e)
        print("--- make sure that file exists ---")

open_file("users.txt", "r")
```

### Reading Line by Line

There are multiple ways to read files line by line:
1. Using `readlines()` - returns a list of all lines
2. Iterating over the file object directly


move file pointer

```python
""" you need to define , r --> read, w--> write, a--> append """
def open_file(file_name, mode='r'):
    # I need to open file for reading ?
    try:
        file_object = open(file_name, mode)
        # read content  into one string
        data = file_object.read() # data ??
        print(data, type(data))
        print("*****************************")
        file_object.seek(0) # reset pointer to the position 0
        data2 = file_object.read()
        print(data2, type(data2), len(data2))
        file_object.close()
    except Exception as e:
        print("--- make sure that file exists ---")

open_file("users.txt", "r")
```

## ✍️ Writing to Files

### Write Mode (`w`)

When opening a file in write mode (`w`):
- If the file doesn't exist, Python creates it
- If the file exists, **it overwrites the existing content**
- The file pointer starts at the beginning


read file content to a list .... line by line ?

```python
""" you need to define , r --> read, w--> write, a--> append """
def read_data_by_lines(file_name, mode='r'):
    # I need to open file for reading ?
    try:
        file_object = open(file_name, mode)
        # read content  into one list
        lines = file_object.readlines()
        print(lines, type(lines))
        file_object.close()
    except Exception as e:
        print(e)
        print("--- make sure that file exists ---")

read_data_by_lines("users.txt", "r")
```

```python
def read_line_by_line(filename, mode='r'):
    try:
        file_object = open(filename, mode)
        for line in file_object:
            print(f"line = {line}")

    except Exception as e :
        return False

read_line_by_line("users.txt", "r")
```

```python
def read_line_by_line(filename, mode='r'):
    try:
        file_object = open(filename, mode)
        for line in file_object:
            line = line.strip("\n")
            fields = line.split(":")
            print(f"id:{fields[0]}, name:{fields[1]}, salary:{fields[2]}")

        file_object.close()

    except Exception as e :
        return False

read_line_by_line("users.txt", "r")
```

## ➕ Appending to Files

### Append Mode (`a`)

When opening a file in append mode (`a`):
- If the file doesn't exist, Python creates it
- If the file exists, new content is added **at the end** of the file
- The file pointer starts at the end


# write data to a file
when open file with w mode, if file not exist python try to  create it,, if file exists --> override content , open file for writing starting the beginning of the file


```python
def write_data_to_file(filename , data_to_write):
    try:
        file_object = open(filename, 'w')
        print(file_object)
        file_object.write(data_to_write)

        file_object.close()

    except Exception as e :
        return False

write_data_to_file("mycv.txt", 'My name is noha')
```

```python
write_data_to_file("mycv.txt", "New data")
```

```python
def write_data_to_file_at_position(filename , data_to_write, bytenumber):
    try:
        file_object = open(filename, 'w')
        file_object.seek(bytenumber)
        file_object.write(data_to_write)
        file_object.close()

    except Exception as e :
        return False

write_data_to_file_at_position("mycv.txt", 'My name is noha', 15)
```

```python
write_data_to_file_at_position("mycv.txt", 'My name is Ghada', -1)
```

## 🔒 Using `with` Statement

The `with` statement is the **recommended way** to work with files in Python. It automatically handles file closing, even if an error occurs, making your code cleaner and safer.


---

## 📝 Summary

**File Modes:**
- `r` - Read mode (default)
- `w` - Write mode (overwrites existing content)
- `a` - Append mode (adds to end of file)

**Key Methods:**
- `read()` - Read entire file as string
- `readlines()` - Read all lines as a list
- `write()` - Write data to file
- `seek(position)` - Move file pointer to specific position
- `close()` - Close the file

**Best Practices:**
- ✅ Always use `with` statement for file operations
- ✅ Handle exceptions when working with files
- ✅ Remember to add `\n` for new lines when writing
- ✅ Close files after use (automatic with `with` statement)


# Append
if file not exists =-> python try to create it,
if exists  --> will open file for writing starting from the end of the file.

```python
def append_data_to_file(filename , data_to_write):
    try:
        file_object = open(filename, 'a')
        print(file_object)
        file_object.write(data_to_write)

        file_object.close()

    except Exception as e :
        return False

append_data_to_file("students.txt", 'My name is noha')
```

```python
append_data_to_file("students.txt", 'My name is Ali')
```

```python
append_data_to_file("students.txt", 'My name is noha\n')
append_data_to_file("students.txt", 'My name is noha\n')
```

# with open

```python
try:
    with open("students.txt", "r") as file_object:
        data = file_object.read()
        print(data, type(data))
except Exception as e:
    print(e)

print(file_object)

```
