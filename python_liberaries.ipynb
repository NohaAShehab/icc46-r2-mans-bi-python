## 📧 Email Validation with `re.match()`

### Email Pattern Breakdown:
```
[A-Za-z0-9._%+-]+  @  [A-Za-z0-9.-]+  \.  [A-Za-z]{2,7}
     ↑                    ↑            ↑        ↑
  username            domain name     dot    top-level domain
```

- `[A-Za-z0-9._%+-]+` - One or more characters (letters, numbers, dots, etc.)
- `@` - Literal @ symbol
- `[A-Za-z0-9.-]+` - Domain name
- `\.` - Escaped dot (literal period)
- `[A-Za-z]{2,7}` - Top-level domain (2-7 letters)

### `re.match()` Function:
- Checks if the **beginning** of the string matches the pattern
- Returns a `Match` object if found, `None` otherwise
- **Important**: `match()` only checks the start of the string, not the entire string!


````python
import re
email_pattern = r"[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,7}"
email  = input("Enter your email address: ")
# check if email valid  --
valid_mail = re.match(email_pattern, email)
print(valid_mail)
if valid_mail:
    print("Email address valid")
else:
    print("Email address invalid")

````

## ⚠️ Limitation of `re.match()`

**Problem**: `re.match()` only checks if the pattern matches at the **start** of the string.

**Example**: 
- `"noha@gmail.com"` ✅ Matches
- `"hello noha@gmail.com"` ❌ Doesn't match (has text before email)
- `"noha@gmail.com extra"` ✅ Matches (but shouldn't!)

For strict validation, we need `re.fullmatch()` instead.


````python
import re
email_pattern = r"[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,7}"
email  = input("Enter your email address: ")
# check if email valid  --
valid_mail = re.match(email_pattern, email)
print(valid_mail)
if valid_mail:
    print("Email address valid")
else:
    print("Email address invalid")

````

## ✅ Using `re.fullmatch()` for Strict Validation

### `re.fullmatch()` Function:
- Ensures the **entire string** matches the pattern
- Returns a `Match` object only if the whole string matches
- **Best practice** for validation (emails, phone numbers, etc.)

### When to Use:
- ✅ **`re.fullmatch()`** - For validation (email, phone, ID formats)
- ✅ **`re.match()`** - For finding patterns at the start of strings
- ✅ **`re.search()`** - For finding patterns anywhere in the string

**Recommendation**: Always use `fullmatch()` for input validation!


- function match-> return with match object if the first part of the string follow patten

````python
import re
email_pattern = r"[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,7}"
email  = input("Enter your email address: ")
# check if email valid  --
valid_mail = re.match(email_pattern, email)
print(valid_mail)
if valid_mail:
    print("Email address valid")
else:
    print("Email address invalid")

````

## 📝 Summary

### Key Differences:

| Function | Checks | Use Case |
|----------|--------|----------|
| `re.match()` | Beginning of string | Finding patterns at start |
| `re.fullmatch()` | Entire string | Input validation |
| `re.search()` | Anywhere in string | Finding patterns in text |

### Best Practices:
1. Use `re.fullmatch()` for validating user input (emails, passwords, etc.)
2. Always use raw strings (`r"pattern"`) for regex patterns
3. Handle `None` return values when validation fails
4. Test your patterns with various inputs before deploying
