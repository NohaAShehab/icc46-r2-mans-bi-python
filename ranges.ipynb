# 🔢 Range Function

<div style="background: linear-gradient(135deg, #667eea 0%, #764ba2 100%); padding: 20px; border-radius: 10px; color: white; margin: 10px 0;">
<h2 style="color: white; margin-top: 0;">📚 What is Range?</h2>
<p>The <code style="background: rgba(255,255,255,0.2); padding: 2px 6px; border-radius: 4px;">range()</code> function generates a sequence of numbers. It's commonly used in loops and is memory-efficient!</p>
</div>

<div style="background: #f0f4ff; padding: 15px; border-left: 4px solid #667eea; margin: 10px 0;">
<h3 style="color: #667eea; margin-top: 0;">✨ Syntax Options:</h3>
<ul>
<li><code style="background: #e8eaff; padding: 2px 6px; border-radius: 4px; color: #764ba2;">range(stop)</code> - starts at 0, goes up to (but not including) stop</li>
<li><code style="background: #e8eaff; padding: 2px 6px; border-radius: 4px; color: #764ba2;">range(start, stop)</code> - starts at start, goes up to (but not including) stop</li>
<li><code style="background: #e8eaff; padding: 2px 6px; border-radius: 4px; color: #764ba2;">range(start, stop, step)</code> - starts at start, increments by step, stops before stop</li>
</ul>
</div>


```python
# range ??
myr = range(10)
```

## 🔄 Range is Iterable

<div style="background: linear-gradient(135deg, #f093fb 0%, #f5576c 100%); padding: 15px; border-radius: 8px; color: white; margin: 10px 0;">
<p style="margin: 0;"><strong>💡 Key Concept:</strong> Range objects are <strong>iterable</strong> - you can loop through them directly without converting to a list first!</p>
</div>

<div style="background: #fff5f5; padding: 12px; border-left: 4px solid #f5576c; margin: 10px 0;">
<p style="margin: 0; color: #721c24;">This is memory-efficient because range doesn't store all values in memory - it generates them on-the-fly!</p>
</div>


```python
# range is iterable object ?

for i in myr:
    print(i)
```

## 📋 Convert Range to List

<div style="background: linear-gradient(135deg, #4facfe 0%, #00f2fe 100%); padding: 15px; border-radius: 8px; color: white; margin: 10px 0;">
<p style="margin: 0;"><strong>🔄 Conversion:</strong> You can convert a range object to a list to see all values at once</p>
</div>

<div style="background: #e6f7ff; padding: 12px; border-left: 4px solid #4facfe; margin: 10px 0;">
<p style="margin: 0; color: #0050b3;"><strong>Note:</strong> Use <code style="background: #bae7ff; padding: 2px 6px; border-radius: 4px;">list(range())</code> when you need to see or manipulate all values</p>
</div>


```python
nums = list(myr)
print(nums)
```

## ⚠️ Range with Start > Stop (No Step)

<div style="background: linear-gradient(135deg, #fa709a 0%, #fee140 100%); padding: 15px; border-radius: 8px; margin: 10px 0;">
<p style="margin: 0; color: #721c24;"><strong>⚠️ Important:</strong> If start is greater than stop without a negative step, range returns <strong>empty</strong></p>
</div>

<div style="background: #fff3cd; padding: 12px; border-left: 4px solid #ffc107; margin: 10px 0;">
<p style="margin: 0; color: #856404;">Example: <code style="background: #ffe69c; padding: 2px 6px; border-radius: 4px;">range(10, 1)</code> returns empty because we can't count forward from 10 to 1</p>
</div>


```python
newr2 = range(10, 1)
print(list(newr2))
```

## 🔙 Range with Negative Step

<div style="background: linear-gradient(135deg, #30cfd0 0%, #330867 100%); padding: 15px; border-radius: 8px; color: white; margin: 10px 0;">
<p style="margin: 0;"><strong>🔙 Counting Backwards:</strong> To count backwards, use a <strong>negative step</strong> value!</p>
</div>

<div style="background: #e0f7fa; padding: 12px; border-left: 4px solid #00acc1; margin: 10px 0;">
<p style="margin: 0; color: #006064;">Example: <code style="background: #b2ebf2; padding: 2px 6px; border-radius: 4px;">range(10, 1, -1)</code> counts from 10 down to 2 (stops before 1)</p>
</div>


```python
newr2 = range(10, 1, -1)
print(list(newr2))
```

## ⚠️ Invalid Range Direction

<div style="background: linear-gradient(135deg, #ff6b6b 0%, #ee5a6f 100%); padding: 15px; border-radius: 8px; color: white; margin: 10px 0;">
<p style="margin: 0;"><strong>❌ Common Mistake:</strong> If step direction doesn't match start/stop direction, range returns empty!</p>
</div>

<div style="background: #ffebee; padding: 12px; border-left: 4px solid #f44336; margin: 10px 0;">
<p style="margin: 0; color: #c62828;">Example: <code style="background: #ffcdd2; padding: 2px 6px; border-radius: 4px;">range(-1, 20, -2)</code> tries to go backwards from -1 to 20, which is impossible!</p>
</div>


```python
newr2 = range(-1, 20, -2 )
print(list(newr2))
```

# 🛑 Loop Control: Break Statement

<div style="background: linear-gradient(135deg, #ff9a9e 0%, #fecfef 100%); padding: 20px; border-radius: 10px; margin: 10px 0;">
<h2 style="color: #2c3e50; margin-top: 0;">🛑 What is Break?</h2>
<p style="color: #2c3e50;"><strong>Break</strong> exits the loop <strong>immediately</strong> when a condition is met - no more iterations!</p>
</div>

<div style="background: #fff3e0; padding: 15px; border-left: 4px solid #ff9800; margin: 10px 0;">
<p style="margin: 0; color: #e65100;"><strong>📝 In this example:</strong> The loop stops early if user enters "quit" - perfect for user-controlled loops!</p>
</div>

<div style="background: #f3e5f5; padding: 12px; border-radius: 6px; margin: 10px 0;">
<p style="margin: 0; color: #6a1b9a;"><strong>💡 Tip:</strong> Break is useful when you want to exit a loop based on a condition, not just when iterations are done</p>
</div>


```python
# I need to stop the loop if action happened
values = []
for i in range(5):
    value = input("please enter a value")
    if value == "quit":
        break
    values.append(value)

print(values)
```

```python
# I need to stop the loop if action happened
values = []
for i in range(5):
    value = input("please enter a value")
    if value == "quit":
        break
    values.append(value)

print(values)
```

```python
# continue
# I need to stop the loop if action happened
values = []
for i in range(5):
    value = input("please enter a value")
    if value == "quit":
        continue # skip the rest of the code for this iteration
    values.append(value)

print(values)
```

# ⏭️ Loop Control: Continue Statement

**Continue** skips the rest of the current iteration and moves to the next one:

In this example, "quit" entries are skipped but the loop continues:


```python
ll = []
while True:
    num = input("please enter a number")
    if num == "quit":
        break
    ll.append(num)
```

# 🔄 While Loop with Break

**While True** creates an infinite loop that runs until `break` is called:

This pattern is useful for user input that continues until a sentinel value:


```python
print(ll)
```

```python
counter = 0
while counter < 5:
    print(counter)
    counter +=1

```
