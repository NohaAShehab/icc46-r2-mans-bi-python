# 🔢 Range Function

The `range()` function generates a sequence of numbers. It's commonly used in loops.

**Syntax:** `range(stop)` or `range(start, stop)` or `range(start, stop, step)`


```python
# range ??
myr = range(10)
```

## 🔄 Range is Iterable

Range objects are **iterable** - you can loop through them directly:


```python
# range is iterable object ?

for i in myr:
    print(i)
```

## 📋 Convert Range to List

You can convert a range object to a list to see all values:


```python
nums = list(myr)
print(nums)
```

## ⚠️ Range with Start > Stop (No Step)

If start is greater than stop without a negative step, range returns empty:


```python
newr2 = range(10, 1)
print(list(newr2))
```

## 🔙 Range with Negative Step

To count backwards, use a **negative step** value:


```python
newr2 = range(10, 1, -1)
print(list(newr2))
```

## ⚠️ Invalid Range Direction

If step direction doesn't match start/stop direction, range returns empty:


```python
newr2 = range(-1, 20, -2 )
print(list(newr2))
```

# 🔁 Using Range in Loops

## When to Use `for` vs `while`?

- **`for` loop**: Use when you know how many iterations you need
- **`while` loop**: Use when you need to loop until a condition is met

Here's a simple example repeating code 5 times:


```python
# code I need to repeat
# when I can use for , while ?

for i in range(5):
    print("Hello World")
```

# 🛑 Loop Control: Break Statement

**Break** exits the loop immediately when a condition is met:

In this example, the loop stops early if user enters "quit":


```python
# I need to stop the loop if action happened
values = []
for i in range(5):
    value = input("please enter a value")
    if value == "quit":
        break
    values.append(value)

print(values)
```

```python
# I need to stop the loop if action happened
values = []
for i in range(5):
    value = input("please enter a value")
    if value == "quit":
        break
    values.append(value)

print(values)
```

```python
# continue
# I need to stop the loop if action happened
values = []
for i in range(5):
    value = input("please enter a value")
    if value == "quit":
        continue # skip the rest of the code for this iteration
    values.append(value)

print(values)
```

# ⏭️ Loop Control: Continue Statement

**Continue** skips the rest of the current iteration and moves to the next one:

In this example, "quit" entries are skipped but the loop continues:


```python
ll = []
while True:
    num = input("please enter a number")
    if num == "quit":
        break
    ll.append(num)
```
