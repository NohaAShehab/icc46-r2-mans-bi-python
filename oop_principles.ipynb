## <span style='color:#1976D2'>📚 Employee Class with Class Methods</span>

<div style='background-color:#E3F2FD; padding:12px; border-radius:8px'>
<p style='color:#1565C0'>Let's start with an Employee class that demonstrates <span style='color:#1976D2'><strong>class variables</strong></span> and <span style='color:#0288D1'><strong>class methods</strong></span>:</p>
</div>


```python
#
class Employee:
    emp_count = 0
    def __init__(self, name, email, salary):
        """
        name, email, salary --> instance variables
        self: object of Employee class
        :param name:
        :param email:
        :param salary:
        """
        self.name = name
        self.email = email
        self.salary = salary
        self.__class__.emp_count += 1


    def display_employee(self):
        print(f"email:{self.name} salary:{self.salary} email:  {self.email}")

    # class method
    @classmethod
    def create_employee(cls):
        return cls("name", "email", 100)


print(Employee.emp_count)
emp1 = Employee("sddd", "", 10000)
emp2 = Employee.create_employee()
emp1.display_employee()
print(Employee.emp_count)
```

## <span style='color:#F57C00'>🧮 Calculating Net Salary</span>

<div style='background-color:#FFF3E0; padding:12px; border-radius:8px'>
<p style='color:#E65100'>Let's create a <span style='color:#FF6F00'><strong>function</strong></span> to calculate net salary (80% of gross salary):</p>
</div>


```python
def cal_net_sal(salary):
    net_salary = salary * 0.8
    return net_salary

print(cal_net_sal(emp1.salary))
```

```python
#
class Employee:
    emp_count = 0
    def __init__(self, name, email, salary):
        """
        name, email, salary --> instance variables
        self: object of Employee class
        :param name:
        :param email:
        :param salary:
        """
        self.name = name
        self.email = email
        self.salary = salary
        self.__class__.emp_count += 1


    def display_employee(self):
        print(f"email:{self.name} salary:{self.salary} email:  {self.email}")

    # class method
    @classmethod
    def create_employee(cls):
        return cls("name", "email", 100)

    def cal_net_sal(salary):
        print(salary)
        net_salary = salary * 0.8
        return net_salary

emp1 = Employee("sddd", "", 10000)
print(emp1.display_employee())
emp1.cal_net_sal()
```

## <span style='color:#D32F2F'>⚠️ The Problem: Methods Without `self`</span>

<div style='background-color:#FFEBEE; padding:12px; border-radius:8px; border-left:4px solid #E53935'>
<p style='color:#C62828'><strong>⚠️ Issue:</strong> If we define a method without <code style='background-color:#FFCDD2; color:#B71C1C; padding:3px 6px; border-radius:3px'>self</code>, it won't work correctly as an instance method.</p>
<p style='color:#D32F2F'>Let's see what happens:</p>
</div>


```python
## Usually first parameter of a method is named 'self'

class Employee:
    emp_count = 0
    def __init__(self, name, email, salary):
        """
        name, email, salary --> instance variables
        self: object of Employee class
        :param name:
        :param email:
        :param salary:
        """
        self.name = name
        self.email = email
        self.salary = salary
        self.__class__.emp_count += 1


    def display_employee(self):
        print(f"email:{self.name} salary:{self.salary} email:  {self.email}")

    # class method
    @classmethod
    def create_employee(cls):
        return cls("name", "email", 100)
    """ this function neither depends on the class nor on the object//instance ,, it is just helper function"""
    @staticmethod
    def cal_net_sal(salary):
        print(salary)
        net_salary = salary * 0.8
        return net_salary

emp1 = Employee("sddd", "", 10000)
emp1.display_employee()
emp1.cal_net_sal()
```

### <span style='color:#FF6F00'>🔍 Basic Inheritance Example</span>

<div style='background-color:#FFF3E0; padding:10px; border-radius:6px'>

```python
class Parent:
    pass

class Child(Parent):
    pass

# inheritance --> is-a relationship
c = Child()
print(isinstance(c, Parent))
```

<p style='color:#1976D2'>If the child doesn't define its own <code style='background-color:#BBDEFB; color:#0D47A1; padding:3px 6px; border-radius:3px'>__init__</code>, it will use the parent's:</p>

</div>


```python
class Employee:
    def __init__(self, name):
        self.name = name

    def display_employee(self):
        print(f"name:{self.name}")

class Manager(Employee):
    pass

mm = Manager("Test")
mm.display_employee()
```

### <span style='color:#00796B'>👔 Manager Inheriting from Employee</span>

<div style='background-color:#E0F2F1; padding:10px; border-radius:6px'>
<p style='color:#00695C'>A <span style='color:#00897B'><strong>Manager</strong></span> is an <span style='color:#009688'><strong>Employee</strong></span>, so it inherits all Employee methods:</p>
</div>


```python
class Employee:
    def __init__(self, name):
        self.name = name

    def display_employee(self):
        print(f"name:{self.name}")

class Manager(Employee):
    def __init__(self, email):
        self.email = email

mm = Manager("Test")
mm.display_employee()
```

### <span style='color:#D32F2F'>⚠️ Overriding `__init__`</span>

<div style='background-color:#FFEBEE; padding:12px; border-radius:8px; border-left:4px solid #E53935'>
<p style='color:#C62828'><strong>⚠️ Problem:</strong> If the child class defines its own <code style='background-color:#FFCDD2; color:#B71C1C; padding:3px 6px; border-radius:3px'>__init__</code>, it <span style='color:#D32F2F'><strong>overrides</strong></span> the parent's.</p>
<p style='color:#E53935'>The parent's <code style='background-color:#FFCDD2; color:#B71C1C; padding:3px 6px; border-radius:3px'>__init__</code> won't be called automatically!</p>
</div>


```python
# explicitly call parent constructor ?
class Employee:
    def __init__(self, name):
        self.name = name

    def display_employee(self):
        print(f"name:{self.name}")

class Manager(Employee):
    def __init__(self, name, email):
        self.email = email
        super().__init__(name)

mm = Manager("Test", "Test@gmail.com")
mm.display_employee()
```

### <span style='color:#00796B'>✅ Using `super()` to Call Parent Constructor</span>

<div style='background-color:#E0F2F1; padding:12px; border-radius:8px; border-left:5px solid #009688'>
<p style='color:#00695C'><strong>✅ Solution:</strong> Use <code style='background-color:#B2DFDB; color:#004D40; padding:3px 6px; border-radius:3px'>super()</code> to explicitly call the parent's <code style='background-color:#B2DFDB; color:#004D40; padding:3px 6px; border-radius:3px'>__init__</code>.</p>
<p style='color:#00897B'>This ensures the parent class is properly initialized:</p>
</div>


```python
# explicitly call parent constructor ?
class Employee:
    def __init__(self, name):
        self.name = name

    def display_employee(self):
        print(f"name:{self.name}")

class Manager(Employee):
    def __init__(self, name, email):
        self.email = email
        super().__init__(name)

    def display_employee(self):
        print(f"email:{self.email}")

mm = Manager("Test", "Test@gmail.com")
mm.display_employee()
```

### <span style='color:#E91E63'>🔄 Method Overriding</span>

<div style='background-color:#FCE4EC; padding:12px; border-radius:8px; border-left:4px solid #E91E63'>
<p style='color:#AD1457'>Child classes can <span style='color:#C2185B'><strong>override</strong></span> parent methods by defining a method with the same name.</p>
<p style='color:#C2185B'>The child's version will be used instead of the parent's:</p>
</div>


```python
# explicitly call parent constructor ?
class Employee:
    def __init__(self, name):
        self.name = name

    def display_employee(self):
        print(f"name:{self.name}")

class Manager(Employee):
    def __init__(self, name, email):
        self.email = email
        super().__init__(name)

    #override
    def display_employee(self):
        print(f"email:{self.email}")
        super().display_employee()

mm = Manager("Test", "Test@gmail.com")
mm.display_employee()
```

### <span style='color:#1976D2'>🔗 Overriding with `super()` Call</span>

<div style='background-color:#E3F2FD; padding:12px; border-radius:8px; border-left:4px solid #2196F3'>
<p style='color:#1565C0'>We can override a method and still call the parent's version using <code style='background-color:#BBDEFB; color:#0D47A1; padding:3px 6px; border-radius:3px'>super()</code>:</p>
<p style='color:#1976D2'>This allows us to <span style='color:#0288D1'><strong>extend</strong></span> parent functionality rather than replace it:</p>
</div>

## <span style='color:#D32F2F; font-size:1.5em'>🔒 Access Modifiers</span>

<div style='background:linear-gradient(135deg, #FFE0B2 0%, #FFCCBC 100%); padding:15px; border-radius:10px; border:2px solid #FF6F00'>
<p style='color:#BF360C; font-size:1.1em'><strong>🎯 Python Access Modifiers:</strong></p>
<ul style='color:#D84315; font-size:1em'>
<li><span style='color:#E65100'><strong>Public:</strong></span> Variable starts with letter (default) - accessible anywhere</li>
<li><span style='color:#F57C00'><strong>Protected:</strong></span> Variable starts with <code style='background-color:#FFF3E0; color:#E65100; padding:2px 5px'>_</code> - should only be accessed within class and subclasses</li>
<li><span style='color:#FF6F00'><strong>Private:</strong></span> Variable starts with <code style='background-color:#FFF3E0; color:#E65100; padding:2px 5px'>__</code> - should only be accessed within the class</li>
</ul>
<p style='color:#E65100'><strong>⚠️ Note:</strong> Python doesn't enforce these - they're conventions!</p>
</div>


"""

object oriented

object contain data => you can apply functionality on.

limit accessibility of this data
access modifiers
    -> public ==> accessed anywhere using object
    -> protected ==> accessed inside the class and derived class
    -> private ==> accessed only inside the class

--> apply validation/ logic on data
--> create readonly properties
"""


# No access modifiers keywords in Python.
- limit accessibilty based variable name
- varaible start with Alpha ===> Public
- variable start with _ ==> protected
- variable start with __ ==> private

### <span style='color:#00796B'>🔧 Getter and Setter Methods</span>

<div style='background-color:#E0F2F1; padding:12px; border-radius:8px; border-left:5px solid #009688'>
<p style='color:#00695C'><strong>✅ Best Practice:</strong> Use <span style='color:#00897B'><strong>getter</strong></span> and <span style='color:#009688'><strong>setter</strong></span> methods to control access to private attributes.</p>
<p style='color:#00897B'>This allows us to add <span style='color:#00796B'><strong>validation</strong></span> and <span style='color:#00695C'><strong>logic</strong></span> when getting/setting values:</p>
</div>


```python
class Employee:
    def __init__(self, name, email, salary):
        self.name= name
        self._email = email
        self.__salary = salary

    def get_salary(self):
        return self.__salary

    def set_salary(self, salary):
        if isinstance(salary, float) or isinstance(salary, int) and salary > 0:
            self.__salary = salary
        else:
            raise TypeError("salary must be float or int")

emp=Employee("noha", "noha@gmail.com", 545454)
emp.set_salary("jdsajhf")
```

### <span style='color:#F57C00'>🧪 Testing Validation</span>

<div style='background-color:#FFF3E0; padding:10px; border-radius:6px'>
<p style='color:#E65100'>The setter method validates the input and raises an error if invalid:</p>
</div>


```python
class Employee:
    def __init__(self, name, email, salary):
        self.name= name
        self._email = email
        self.__salary = salary

    def get_salary(self):
        return self.__salary

    def set_salary(self, salary):
        if isinstance(salary, float) or isinstance(salary, int) and salary > 0:
            self.__salary = salary
        else:
            raise TypeError("salary must be float or int")

emp=Employee("noha", "noha@gmail.com", "545454")
```

### <span style='color:#1976D2'>🔍 Validation in `__init__`</span>

<div style='background-color:#E3F2FD; padding:10px; border-radius:6px'>
<p style='color:#1565C0'>We should also validate in <code style='background-color:#BBDEFB; color:#0D47A1; padding:3px 6px; border-radius:3px'>__init__</code> to catch errors during object creation:</p>
</div>


```python
class Employee:
    def __init__(self, name, email, salary):
        self.name= name
        self._email = email
        if isinstance(salary, float) or isinstance(salary, int) and salary > 0:
            self.__salary = salary
        else:
            raise TypeError("salary must be float or int")

    def get_salary(self):
        return self.__salary

    def set_salary(self, salary):
        if isinstance(salary, float) or isinstance(salary, int) and salary > 0:
            self.__salary = salary
        else:
            raise TypeError("salary must be float or int")

emp=Employee("noha", "noha@gmail.com", "545454")
```

### <span style='color:#7B1FA2'>🔄 Using Setter in `__init__`</span>

<div style='background-color:#F3E5F5; padding:10px; border-radius:6px'>
<p style='color:#6A1B9A'>A better approach: call the setter method from <code style='background-color:#E1BEE7; color:#4A148C; padding:3px 6px; border-radius:3px'>__init__</code> to avoid code duplication:</p>
</div>


```python
class Employee:
    def __init__(self, name, email, salary):
        self.name= name
        self._email = email
        self.set_salary(salary)

    def get_salary(self):
        return self.__salary

    def set_salary(self, salary):
        if isinstance(salary, float) or isinstance(salary, int) and salary > 0:
            self.__salary = salary
        else:
            raise TypeError("salary must be float or int")

emp=Employee("noha", "noha@gmail.com", "545454")
```

### <span style='color:#00796B'>🔧 Using Setter (Duplicate)</span>

<div style='background-color:#E0F2F1; padding:10px; border-radius:6px'>
<p style='color:#00695C'>Same approach - reusing the setter method for validation:</p>
</div>


```python
class Employee:
    def __init__(self, name, email, salary):
        self.name= name
        self._email = email
        self.set_salary(salary)

    def get_salary(self):
        return self.__salary

    def set_salary(self, salary):
        if isinstance(salary, float) or isinstance(salary, int) and salary > 0:
            self.__salary = salary
        else:
            raise TypeError("salary must be float or int")

emp=Employee("noha", "noha@gmail.com", "545454")
```

## <span style='color:#D32F2F; font-size:1.5em'>✨ Properties</span>

<div style='background:linear-gradient(135deg, #C5E1A5 0%, #AED581 100%); padding:15px; border-radius:10px; border:2px solid #689F38'>
<p style='color:#33691E; font-size:1.1em'><strong>🎯 Properties</strong> allow us to use <code style='background-color:#E8F5E9; color:#2E7D32; padding:3px 6px; border-radius:3px'>@property</code> decorator to create <span style='color:#558B2F'><strong>readable attributes</strong></span> that can have custom logic.</p>
<p style='color:#388E3C'>They make getters and setters look like regular attribute access!</p>

# property decorator

### <span style='color:#1976D2'>📊 Employee with Getter Method</span>

<div style='background-color:#E3F2FD; padding:10px; border-radius:6px'>
<p style='color:#1565C0'>First, let's see the traditional getter method approach:</p>
</div>


```python
class Employee:
    def __init__(self, name, email, salary):
        self.name= name
        self._email = email
        self.set_salary(salary)

    def get_salary(self):
        return self.__salary

    def set_salary(self, salary):
        if isinstance(salary, float) or isinstance(salary, int) and salary > 0:
            self.__salary = salary
        else:
            raise TypeError("salary must be float or int")

emp=Employee("noha", "noha@gmail.com", "545454")
print(emp.get_salary())
```

### <span style='color:#E91E63'>✨ Using `@property` Decorator</span>

<div style='background-color:#FCE4EC; padding:12px; border-radius:8px; border-left:5px solid #E91E63'>
<p style='color:#AD1457'>The <code style='background-color:#F8BBD0; color:#880E4F; padding:3px 6px; border-radius:3px'>@property</code> decorator allows us to access methods like attributes!</p>
<p style='color:#C2185B'>Now we can use <code style='background-color:#F8BBD0; color:#880E4F; padding:3px 6px; border-radius:3px'>emp.salary</code> instead of <code style='background-color:#F8BBD0; color:#880E4F; padding:3px 6px; border-radius:3px'>emp.get_salary()</code>:</p>
</div>


```python
class Employee:
    def __init__(self, name, email, salary):
        self.name= name
        self._email = email
        self.set_salary(salary)

    @property
    def salary(self):
        return self.__salary*.8

    def set_salary(self, salary):
        if isinstance(salary, float) or isinstance(salary, int) and salary > 0:
            self.__salary = salary
        else:
            raise TypeError("salary must be float or int")

emp=Employee("noha", "noha@gmail.com", 78451)
print(emp.__dict__)
print(emp.salary)
```

### <span style='color:#1976D2'>✏️ Adding Setter with `@property.setter`</span>

<div style='background-color:#E3F2FD; padding:12px; border-radius:8px; border-left:4px solid #2196F3'>
<p style='color:#1565C0'>We can add a setter using <code style='background-color:#BBDEFB; color:#0D47A1; padding:3px 6px; border-radius:3px'>@property_name.setter</code> decorator.</p>
<p style='color:#1976D2'>Now we can set values using <code style='background-color:#BBDEFB; color:#0D47A1; padding:3px 6px; border-radius:3px'>emp.salary = value</code>:</p>
</div>


```python
class Employee:
    def __init__(self, name, email, salary):
        self.name= name
        self._email = email
        self.__salary = salary
    @property
    def salary(self):
        return self.__salary*.8

    @salary.setter
    def salary(self, salary):
        if isinstance(salary, float) or isinstance(salary, int) and salary > 0:
            self.__salary = salary
        else:
            raise TypeError("salary must be float or int")

emp=Employee("noha", "noha@gmail.com", 78451)
emp.salary = 1000
print(emp.__dict__)
print(emp.salary)
```

### <span style='color:#D32F2F'>⚠️ Property Initialization Issue</span>

<div style='background-color:#FFEBEE; padding:12px; border-radius:8px; border-left:4px solid #E53935'>
<p style='color:#C62828'><strong>⚠️ Problem:</strong> Don't set a property in <code style='background-color:#FFCDD2; color:#B71C1C; padding:3px 6px; border-radius:3px'>__init__</code> if it has a setter!</p>
<p style='color:#D32F2F'>This can cause issues because the property setter will be called, which might modify the value.</p>
</div>


```python
class Employee:
    def __init__(self, name, email, salary):
        self.name= name
        self._email = email
        self.__salary = salary
        self.salary = salary

    @property
    def salary(self):
        return self.__salary*.8

    @salary.setter
    def salary(self, salary):
        if isinstance(salary, float) or isinstance(salary, int) and salary > 0:
            self.__salary = salary
        else:
            raise TypeError("salary must be float or int")

emp=Employee("noha", "noha@gmail.com", 78451)
emp.salary = 1000
print(emp.__dict__)
print(emp.salary)
```

### <span style='color:#7B1FA2'>📊 Multiple Properties</span>

<div style='background-color:#F3E5F5; padding:10px; border-radius:6px'>
<p style='color:#6A1B9A'>A class can have <span style='color:#9C27B0'><strong>multiple properties</strong></span>, each with its own getter and setter:</p>
</div>


```python
class Employee:
    def __init__(self, name, email, salary):
        self.name= name
        self._email = email
        self.__salary = salary
        self.salary = salary

    @property
    def commission(self):
        return  self.__salary * .5
    @property
    def salary(self):
        return self.__salary*.8

    @salary.setter
    def salary(self, salary):
        if isinstance(salary, float) or isinstance(salary, int) and salary > 0:
            self.__salary = salary
        else:
            raise TypeError("salary must be float or int")

emp=Employee("noha", "noha@gmail.com", 78451)
emp.salary = 1000
print(emp.__dict__)
print(emp.salary)
print(emp.commission)
```
