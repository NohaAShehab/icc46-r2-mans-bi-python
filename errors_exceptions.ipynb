## 🛡️ Basic Exception Handling

### Using `try-except` Blocks

The `try-except` block allows you to catch and handle exceptions gracefully, preventing your program from crashing.


# exception handling

### Catching Specific Exceptions

You can catch specific exception types to handle different errors differently. This provides more precise error handling.


```python
try:
    num = int(input("Enter a number: "))
except:
    pass
```

```python
try:
    num = int(input("Enter a number: "))
except:
    print("error happened")
```

```python
try:
    num = int(input("Enter a number: "))
except Exception as e:
    print(f"Error happened {e}")
```

```python
try:
    num = int(input("Enter a number: "))
except Exception as e:
    print(f"Error happened {e}")
    num = 0
    exit
```

### The `else` Block

The `else` block is executed only if no exceptions occur in the `try` block. It's useful for code that should run only when everything succeeds.


```python
try:
    num = int(input("Enter a number: "))
    num2 = int(input("Enter a number: "))
    res = num/num2
    print(f"res = {res}")
except Exception as e:
    print(f"Error happened {e}")
    num = 0
```

### The `finally` Block

The `finally` block is **always** executed, regardless of whether an exception occurred or not. It's perfect for cleanup operations like closing files or releasing resources.


```python
try:
    num = int(input("Enter a number: "))
    num2 = int(input("Enter a number: "))
    res = num/num2
    print(f"res = {res}")
except Exception as e:
    print(f"Error happened {e}")
```

```python
try:
    num = int(input("Enter a number: "))
    num2 = int(input("Enter a number: "))
    res = num/num2
    print(f"res = {res}")
except Exception as e:
    print(f"Error happened {e}")
```

## 🚀 Raising Exceptions

Sometimes you need to raise exceptions yourself to signal that something went wrong in your code. This is useful for input validation and enforcing business rules.


```python
try:
    num1 = int(input("Enter a number: "))
    num2 = int(input("Enter a number: "))
    res = num/num2
    print(f"res = {res}")
except ValueError as ve:
    print(f"Error happened {ve}")
    print("num1, num2 must be integers, you enterd string")
    num1 = 0
    num2 =0
    res =0

except ZeroDivisionError as ze:
    print(f"Error happened {ze}")
    print("num2 = zero which is not correct, assume num2=1 ")
    num2=1
    res = num1

except Exception as e:
    print(f"Error happened {e}")
print("operation completed successfully")
```
