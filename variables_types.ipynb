> **💡 Interesting Fact:** Boolean and Integer Comparison

`True == 1` evaluates to `True` because Python treats `True` as `1` in numeric contexts:

```python
True == 1    # True
True == 1.0  # True
False == 0   # True
```

However, they are still different types:
```python
type(True) == type(1)  # False
```


---

> **⚠️ Important:** `True` is only equal to `1`!

`True` is not equal to any other number:

```python
True == 1    # True ✅
True == 100  # False ❌
True == 0    # False ❌
```


---

> **⚠️ Type Mismatch:** String vs Integer

A string `"10"` is **not equal** to the integer `10`. They are different types:

```python
"10" == 10  # False ❌
```

**To compare, convert one to match the other:**
```python
int("10") == 10  # True ✅
"10" == str(10)  # True ✅
```

### `and` Operator Behavior

**How `and` works:**
1. Evaluates left side first
2. If left is **falsy** → returns left value (stops here)
3. If left is **truthy** → evaluates right side and returns right value

**Example:**
```python
"ahmed" and 'iti'  # Returns 'iti'
# Since "ahmed" is truthy, it evaluates 'iti' and returns it
```


---

### 🔄 Converting to Boolean

The `bool()` function converts any value to a boolean:

```python
bool("ahmed")  # True (non-empty string)
bool("")       # False (empty string)
bool(0)        # False
bool(1)        # True
bool(None)     # False
```

**Rule:** Non-empty strings, non-zero numbers, and non-empty collections are `True`.

> **💡 Example:** Using Truthiness in Conditionals

An empty string `""` is falsy, so the `else` block executes:

```python
message = ""
if message:  # False (empty string is falsy)
    print("-- found")
else:
    print("-- not found")  # This executes
```

### `or` Operator Behavior

**How `or` works:**
1. Evaluates left side first
2. If left is **truthy** → returns left value (stops here - short-circuit)
3. If left is **falsy** → evaluates right side and returns right value

**Example:**
```python
"ahmed" or 'iti'  # Returns "ahmed"
# Since "ahmed" is truthy, it returns immediately (short-circuit)
```


````python
numm = 10
print(type(numm))
numm = float(numm)
print(type(numm))
print(numm)
````
