# 🔢 Sets

<div style="background: linear-gradient(135deg, #f093fb 0%, #f5576c 100%); padding: 25px; border-radius: 12px; color: white; margin: 15px 0; box-shadow: 0 4px 6px rgba(0,0,0,0.1);">
<h2 style="color: white; margin-top: 0; font-size: 1.8em;">🔢 What are Sets?</h2>
<p style="font-size: 1.1em; margin-bottom: 20px;">Sets are collections of <strong>unique</strong> elements with special characteristics!</p>

<div style="background: rgba(255,255,255,0.2); padding: 15px; border-radius: 8px; margin: 10px 0;">
<h3 style="color: white; margin-top: 0;">✨ Key Characteristics:</h3>
<ul style="margin: 5px 0; padding-left: 20px;">
<li>🚫 <strong>No Duplicates</strong> - automatically removes duplicate values</li>
<li>🔀 <strong>Unordered</strong> - no guaranteed order of elements</li>
<li>🚫 <strong>No Indexing</strong> - cannot access elements by index</li>
<li>✅ <strong>Mutable</strong> - can add/remove elements</li>
</ul>
</div>
</div>

# 📚 Dictionaries

<div style="background: linear-gradient(135deg, #667eea 0%, #764ba2 100%); padding: 25px; border-radius: 12px; color: white; margin: 15px 0; box-shadow: 0 4px 6px rgba(0,0,0,0.1);">
<h2 style="color: white; margin-top: 0; font-size: 1.8em;">📚 What are Dictionaries?</h2>
<p style="font-size: 1.1em; margin-bottom: 20px;">Dictionaries store data as <strong>key-value pairs</strong> - making your data labeled and meaningful!</p>

<div style="background: rgba(255,255,255,0.2); padding: 15px; border-radius: 8px; margin: 10px 0;">
<h3 style="color: white; margin-top: 0;">✨ Key Features:</h3>
<ul style="margin: 5px 0; padding-left: 20px;">
<li>🔑 <strong>Unique Keys</strong> - each key appears only once</li>
<li>📊 <strong>Ordered</strong> - maintains insertion order (Python 3.7+)</li>
<li>✅ <strong>Mutable</strong> - can add, update, and remove items</li>
<li>⚡ <strong>Fast Lookup</strong> - uses hashing for quick access</li>
</ul>
</div>

<div style="background: rgba(255,255,255,0.2); padding: 15px; border-radius: 8px; margin: 10px 0;">
<h3 style="color: white; margin-top: 0;">💡 Syntax:</h3>
<p style="margin: 5px 0;"><code style="background: rgba(255,255,255,0.2); padding: 2px 6px; border-radius: 4px;">dict = {"key": "value", "key2": "value2"}</code></p>
</div>
</div>


## 📖 Accessing Dictionary Elements

<div style="background: linear-gradient(135deg, #30cfd0 0%, #330867 100%); padding: 15px; border-radius: 8px; color: white; margin: 10px 0;">
<p style="margin: 0;"><strong>📖 Access by Key:</strong> Use square brackets with the key to access values</p>
</div>

<div style="background: #e0f7fa; padding: 12px; border-left: 4px solid #00acc1; margin: 10px 0;">
<p style="margin: 0; color: #006064;">Example: <code style="background: #b2ebf2; padding: 2px 6px; border-radius: 4px;">info["name"]</code> gets the value associated with the "name" key</p>
</div>


## ✏️ Updating Dictionary Values

<div style="background: linear-gradient(135deg, #fa709a 0%, #fee140 100%); padding: 15px; border-radius: 8px; margin: 10px 0;">
<p style="margin: 0; color: #2c3e50;"><strong>✏️ Update:</strong> Assign a new value to an existing key to update it</p>
</div>

<div style="background: #fff3cd; padding: 12px; border-left: 4px solid #ffc107; margin: 10px 0;">
<p style="margin: 0; color: #856404;">Dictionaries are <strong>mutable</strong> - you can modify values in-place!</p>
</div>


## ➕ Adding New Key-Value Pairs

<div style="background: linear-gradient(135deg, #a8edea 0%, #fed6e3 100%); padding: 15px; border-radius: 8px; margin: 10px 0;">
<p style="margin: 0; color: #2c3e50;"><strong>➕ Insert:</strong> Assign a value to a new key to add it to the dictionary</p>
</div>

<div style="background: #f3e5f5; padding: 12px; border-left: 4px solid #9c27b0; margin: 10px 0;">
<p style="margin: 0; color: #4a148c;">If the key doesn't exist, it will be created automatically!</p>
</div>


## ⚠️ Dictionary Keys Must Be Hashable

<div style="background: linear-gradient(135deg, #ff6b6b 0%, #ee5a6f 100%); padding: 15px; border-radius: 8px; color: white; margin: 10px 0;">
<p style="margin: 0;"><strong>⚠️ TypeError:</strong> Dictionary keys must be <strong>hashable</strong> (immutable) types!</p>
</div>

<div style="background: #ffebee; padding: 12px; border-left: 4px solid #f44336; margin: 10px 0;">
<p style="margin: 0; color: #c62828;"><strong>❌ Cannot Use:</strong> Lists (mutable, unhashable)</p>
<p style="margin: 5px 0 0 0; color: #c62828;"><strong>✅ Can Use:</strong> Strings, numbers, tuples (immutable, hashable)</p>
</div>
</div>


## ✅ Using Tuples as Dictionary Keys

<div style="background: linear-gradient(135deg, #4facfe 0%, #00f2fe 100%); padding: 15px; border-radius: 8px; color: white; margin: 10px 0;">
<p style="margin: 0;"><strong>✅ Tuples Work:</strong> Since tuples are immutable, they can be used as dictionary keys!</p>
</div>

<div style="background: #e6f7ff; padding: 12px; border-left: 4px solid #4facfe; margin: 10px 0;">
<p style="margin: 0; color: #0050b3;">This is useful for creating composite keys or lookup tables!</p>
</div>


## 🔑 Unique Keys Rule

<div style="background: linear-gradient(135deg, #ffecd2 0%, #fcb69f 100%); padding: 15px; border-radius: 8px; margin: 10px 0;">
<p style="margin: 0; color: #2c3e50;"><strong>🔑 Important:</strong> If you use the same key twice, the <strong>last value</strong> overwrites the previous one!</p>
</div>

<div style="background: #fff3e0; padding: 12px; border-left: 4px solid #ff9800; margin: 10px 0;">
<p style="margin: 0; color: #e65100;">Dictionary keys must be unique - duplicate keys are not allowed!</p>
</div>


```python
# datatype
# set ??
# no duplication
# un-ordered
# no reference for the item
mys = {"iti", "mansoura", 2423, "iti", True, "iti", True, "new"}
print(mys)
```

<h2 style="color: white; margin-top: 0; font-size: 1.8em;">🔍 Useful Dictionary Methods</h2>
<p style="font-size: 1.1em; margin-bottom: 20px;">Dictionaries have built-in methods to access keys, values, and items!</p>
