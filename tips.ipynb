## 🏷️ Type Hints & Annotations

### Basic Function Without Type Hints

Python is dynamically typed, meaning you can pass any type to a function. However, this can lead to runtime errors if the wrong types are used.


```python
def sum_num(num1, num2):
    print(f"num1={num1}, num2={num2}")
    res = num1 + num2
    return res
```

```python
sum_num(2,34)
```

```python
sum_num("iti", "BI")
```

### Adding Type Hints

Type hints provide **documentation** and help IDEs provide better autocomplete, but they don't enforce type checking at runtime. Python still allows any type to be passed.


```python
sum_num(324, "rewrt")
```

```python
# define inputs datatype
"""
 void sum_num(int num1, int num2) {}
"""

def sum_num(num1: int, num2: int) -> int:
    print(f"num1={num1}, num2={num2}")
    res = num1 + num2
    return res

sum_num(3,23)
```

## ✅ Input Validation

### Manual Type Checking

Since type hints don't enforce runtime checking, you need to **validate inputs yourself** if you want to ensure correct types are passed to your functions.


```python
sum_num("iti", "BI")
```

### Using `isinstance()` for Type Checking

The `isinstance()` function is the recommended way to check types in Python. It's more flexible than `type()` and handles inheritance correctly.


```python
sum_num(324, "rewrt")
```

## 📝 Summary

**Type Hints:**
- Provide documentation and IDE support
- Use syntax: `def function(param: type) -> return_type:`
- **Do NOT enforce runtime type checking**

**Input Validation:**
- ✅ Always validate inputs if type safety is important
- ✅ Use `isinstance()` instead of `type()` for type checking
- ✅ Provide clear error messages when validation fails

**Best Practices:**
- Use type hints for better code documentation
- Validate inputs in critical functions
- Write clear docstrings explaining function behavior
- Handle type errors gracefully with informative messages


```python
def sum_num(num1: int, num2: int) -> int:
    if type(num1) == int and type(num2) == int:
        print(f"num1={num1}, num2={num2}")
        res = num1 + num2
        return res
    else:
        print(f"num1={num1}, num2={num2}")
        print("num1 and num2 must be integers")

sum_num(3,23)
```

```python
sum_num("iti", "BI")
sum_num(324, "rewrt")
```
