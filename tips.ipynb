```python
def sum_num(num1: int, num2: int) -> int:
    if type(num1) == int and type(num2) == int:
        print(f"num1={num1}, num2={num2}")
        res = num1 + num2
        return res
    else:
        print(f"num1={num1}, num2={num2}")
        print("num1 and num2 must be integers")

sum_num(3,23)
```
