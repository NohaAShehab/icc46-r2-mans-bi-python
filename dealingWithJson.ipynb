# 📄 JSON File Operations with Python

This notebook demonstrates how to read and write JSON (JavaScript Object Notation) files in Python.

## Topics Covered:
- Reading JSON files
- Writing JSON files
- Appending data to JSON arrays
- Pretty printing JSON with indentation
- Error handling for file operations


```python
import json

""""""
```

## 📖 Reading and Writing JSON Files

### Steps:
1. Open the JSON file in read mode (`"r"`)
2. Use `json.load()` to parse the JSON data into Python objects
3. Modify the data (e.g., append new items)
4. Open the file in write mode (`"w"`)
5. Use `json.dump()` to write the data back

**Note**: Always use `try-except` blocks to handle potential errors like file not found or invalid JSON format.


```python
try:
    with open("students.json", "r") as file_object:
        students = json.load(file_object)

    info = {"id":3, "name":"noha", "track": "bi"}
    students.append(info)

    with open("students.json", "w") as file_object:
        json.dump(students, file_object)

except Exception as e:
    print(e)

```

## 🎨 Pretty Printing JSON

By default, `json.dump()` writes JSON in a compact format (all on one line).

Use the `indent` parameter to format JSON in a readable way:
- `indent=4` adds 4 spaces for each indentation level
- Makes the JSON file human-readable
- Easier to debug and edit manually


```python
try:
    with open("students.json", "r") as file_object:
        students = json.load(file_object)

    info = {"id":3, "name":"noha", "track": "bi"}
    students.append(info)

    with open("students.json", "w") as file_object:
        json.dump(students, file_object, indent=4)

except Exception as e:
    print(e)
```
