```python
try:
    with open("students.json", "r") as file_object:
        students = json.load(file_object)

    info = {"id":3, "name":"noha", "track": "bi"}
    students.append(info)

    with open("students.json", "w") as file_object:
        json.dump(students, file_object)

except Exception as e:
    print(e)

```
