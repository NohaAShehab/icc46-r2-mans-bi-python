# 🔧 5. Functions with Mandatory Arguments

<div style="background: linear-gradient(135deg, #ff6b6b 0%, #ee5a6f 100%); padding: 25px; border-radius: 12px; color: white; margin: 15px 0; box-shadow: 0 4px 6px rgba(0,0,0,0.1);">
<h2 style="color: white; margin-top: 0; font-size: 1.8em;">🔧 Mandatory Arguments</h2>
<p style="font-size: 1.1em; margin-bottom: 20px;">Functions can accept <strong>parameters</strong> (arguments) that must be provided when calling!</p>

<div style="background: rgba(255,255,255,0.2); padding: 15px; border-radius: 8px; margin: 10px 0;">
<h3 style="color: white; margin-top: 0;">✅ Syntax:</h3>
<p style="margin: 5px 0;"><code style="background: rgba(255,255,255,0.2); padding: 2px 6px; border-radius: 4px;">def function_name(param1, param2):</code></p>
</div>

<div style="background: rgba(255,255,255,0.2); padding: 15px; border-radius: 8px; margin: 10px 0;">
<h3 style="color: white; margin-top: 0;">⚠️ Important:</h3>
<p style="margin: 5px 0;">All mandatory arguments must be provided - missing arguments will cause a <strong>TypeError</strong>!</p>
</div>
</div>

```python
# use paramenters ??

def sum_num(num1, num2):
    return num1 + num2

print(sum_num(1, 2))
```

<div style="background: linear-gradient(135deg, #ff6b6b 0%, #ee5a6f 100%); padding: 15px; border-radius: 8px; color: white; margin: 10px 0;">
<p style="margin: 0;"><strong>❌ TypeError:</strong> Calling a function without all required arguments raises an error!</p>
</div>

<div style="background: #ffebee; padding: 12px; border-left: 4px solid #f44336; margin: 10px 0;">
<p style="margin: 0; color: #c62828;">Always provide all mandatory arguments when calling a function!</p>
</div>


```python
sum_num(33)
```

# 🎛️ 6. Functions with Optional Arguments

<div style="background: linear-gradient(135deg, #667eea 0%, #764ba2 100%); padding: 25px; border-radius: 12px; color: white; margin: 15px 0; box-shadow: 0 4px 6px rgba(0,0,0,0.1);">
<h2 style="color: white; margin-top: 0; font-size: 1.8em;">🎛️ Default Parameters</h2>
<p style="font-size: 1.1em; margin-bottom: 20px;">Functions can have <strong>default values</strong> for parameters - making them optional!</p>

<div style="background: rgba(255,255,255,0.2); padding: 15px; border-radius: 8px; margin: 10px 0;">
<h3 style="color: white; margin-top: 0;">✅ Syntax:</h3>
<p style="margin: 5px 0;"><code style="background: rgba(255,255,255,0.2); padding: 2px 6px; border-radius: 4px;">def function_name(param1, param2=default_value):</code></p>
</div>

<div style="background: rgba(255,255,255,0.2); padding: 15px; border-radius: 8px; margin: 10px 0;">
<h3 style="color: white; margin-top: 0;">💡 Benefits:</h3>
<ul style="margin: 5px 0; padding-left: 20px;">
<li>Flexibility - can call with or without optional arguments</li>
<li>Backward compatibility - add new features without breaking old code</li>
</ul>
</div>
</div>

```python

def sum_num(num1, num2=10):
    print(f"num1={num1}, num2={num2}")
    return num1 + num2

print(sum_num(1, 2))
```

## ⚠️ Parameter Order Rule

<div style="background: linear-gradient(135deg, #ff6b6b 0%, #ee5a6f 100%); padding: 15px; border-radius: 8px; color: white; margin: 10px 0;">
<p style="margin: 0;"><strong>⚠️ SyntaxError:</strong> Parameters with defaults must come <strong>after</strong> parameters without defaults!</p>
</div>

<div style="background: #ffebee; padding: 12px; border-left: 4px solid #f44336; margin: 10px 0;">
<p style="margin: 0; color: #c62828;"><strong>❌ Wrong:</strong> <code style="background: #ffcdd2; padding: 2px 6px; border-radius: 4px;">def func(param1=10, param2):</code></p>
<p style="margin: 5px 0 0 0; color: #c62828;"><strong>✅ Correct:</strong> <code style="background: #ffcdd2; padding: 2px 6px; border-radius: 4px;">def func(param1, param2=10):</code></p>
</div>

</div>

<div style="background: #e6f7ff; padding: 15px; border-radius: 8px; margin: 10px 0;">

<div style="background: linear-gradient(135deg, #a8edea 0%, #fed6e3 100%); padding: 15px; border-radius: 8px; margin: 10px 0;">
<p style="margin: 0; color: #2c3e50;"><strong>🏷️ Named Arguments:</strong> You can pass arguments by name, regardless of order!</p>

<div style="background: #f3e5f5; padding: 12px; border-left: 4px solid #9c27b0; margin: 10px 0;">
<p style="margin: 0; color: #4a148c;">This makes function calls more readable and allows you to skip optional parameters!</p>

## ❌ Error: Too Many Arguments

<div style="background: linear-gradient(135deg, #ff6b6b 0%, #ee5a6f 100%); padding: 15px; border-radius: 8px; color: white; margin: 10px 0;">
<p style="margin: 0;"><strong>❌ TypeError:</strong> Functions with fixed parameters can't accept extra arguments!</p>

<div style="background: #ffebee; padding: 12px; border-left: 4px solid #f44336; margin: 10px 0;">
<p style="margin: 0; color: #c62828;">Use <code style="background: #ffcdd2; padding: 2px 6px; border-radius: 4px;">*args</code> to accept variable number of arguments!</p>

## ✨ *args - Variable Positional Arguments

<div style="background: linear-gradient(135deg, #4facfe 0%, #00f2fe 100%); padding: 20px; border-radius: 10px; color: white; margin: 10px 0;">
<h2 style="color: white; margin-top: 0;">✨ *args Explained</h2>
<p style="margin: 0;"><code style="background: rgba(255,255,255,0.2); padding: 2px 6px; border-radius: 4px;">*args</code> collects extra positional arguments into a <strong>tuple</strong>!</p>
</div>

<div style="background: #e6f7ff; padding: 15px; border-left: 4px solid #4facfe; margin: 10px 0;">
<p style="margin: 0; color: #0050b3;"><strong>💡 Key Points:</strong></p>
<ul style="margin: 5px 0; padding-left: 20px; color: #0050b3;">
<li>Accepts <strong>zero or more</strong> arguments</li>
<li>Arguments are packed into a tuple</li>
<li>The name "args" is a convention - you can use any name</li>
</ul>
</div>


## 🔀 Combining *args with Default Parameters

<div style="background: linear-gradient(135deg, #30cfd0 0%, #330867 100%); padding: 15px; border-radius: 8px; color: white; margin: 10px 0;">
<p style="margin: 0;"><strong>🔀 Order Matters:</strong> When combining <code style="background: rgba(255,255,255,0.2); padding: 2px 6px; border-radius: 4px;">*args</code> with default parameters, order is important!</p>
</div>

<div style="background: #e0f7fa; padding: 12px; border-left: 4px solid #00acc1; margin: 10px 0;">
<p style="margin: 0; color: #006064;"><strong>✅ Correct Order:</strong> <code style="background: #b2ebf2; padding: 2px 6px; border-radius: 4px;">def func(*args, name="default"):</code></p>
<p style="margin: 5px 0 0 0; color: #006064;"><strong>⚠️ Note:</strong> Default parameters after <code style="background: #b2ebf2; padding: 2px 6px; border-radius: 4px;">*args</code> must be called by name!</p>
</div>


## 📍 *args Before Default Parameters

<div style="background: linear-gradient(135deg, #a8edea 0%, #fed6e3 100%); padding: 15px; border-radius: 8px; margin: 10px 0;">
<p style="margin: 0; color: #2c3e50;"><strong>📍 Alternative Order:</strong> You can put <code style="background: rgba(255,255,255,0.5); padding: 2px 6px; border-radius: 4px;">*args</code> after default parameters, but the first positional argument will fill the default!</p>
</div>

<div style="background: #f3e5f5; padding: 12px; border-left: 4px solid #9c27b0; margin: 10px 0;">
<p style="margin: 0; color: #4a148c;">In this case, the first argument goes to the default parameter, and remaining arguments go to <code style="background: #e1bee7; padding: 2px 6px; border-radius: 4px;">*args</code>!</p>
</div>

# 🌟 7. Variable Arguments (*args and **kwargs)

<div style="background: linear-gradient(135deg, #667eea 0%, #764ba2 100%); padding: 25px; border-radius: 12px; color: white; margin: 15px 0; box-shadow: 0 4px 6px rgba(0,0,0,0.1);">
<h2 style="color: white; margin-top: 0; font-size: 1.8em;">🌟 Variable Arguments</h2>
<p style="font-size: 1.1em; margin-bottom: 20px;">Sometimes you need functions that accept an <strong>unknown number</strong> of arguments!</p>

<div style="background: rgba(255,255,255,0.2); padding: 15px; border-radius: 8px; margin: 10px 0;">
<h3 style="color: white; margin-top: 0;">✨ Two Types:</h3>
<ul style="margin: 5px 0; padding-left: 20px;">
<li><code style="background: rgba(255,255,255,0.2); padding: 2px 6px; border-radius: 4px;">*args</code> - for positional arguments (packed as tuple)</li>
<li><code style="background: rgba(255,255,255,0.2); padding: 2px 6px; border-radius: 4px;">**kwargs</code> - for keyword arguments (packed as dictionary)</li>
</ul>
</div>
</div>

## 📚 **kwargs - Keyword Arguments Dictionary

<div style="background: linear-gradient(135deg, #fa709a 0%, #fee140 100%); padding: 20px; border-radius: 10px; margin: 10px 0;">
<h2 style="color: #2c3e50; margin-top: 0;">📚 **kwargs Explained</h2>
<p style="color: #2c3e50; margin-bottom: 0;"><code style="background: rgba(255,255,255,0.5); padding: 2px 6px; border-radius: 4px;">**kwargs</code> collects keyword arguments into a <strong>dictionary</strong>!</p>
</div>

<div style="background: #fff3cd; padding: 15px; border-left: 4px solid #ffc107; margin: 10px 0;">
<p style="margin: 0; color: #856404;"><strong>💡 Use Cases:</strong> Perfect for functions that need flexible configuration options!</p>
</div>


```python
def ask_for_students(students):
    print(students)

ask_for_students("Ahmed")
```

```python
ask_for_students("Ahmed","Mohamed")
```

```python
# I need to say that function may accept zero or more or arguemnt ??
def ask_for_students(*students): # * zero or more
    print(students) # tuple

ask_for_students("Ahmed")
ask_for_students("Ahmed","Mohamed")
ask_for_students()
```

```python
def ask_for_students(*students, name="Ahmed"):
    print(students, name)

ask_for_students("Ahmed",)
```

```python
def ask_for_students( name="Ahmed", *students):
    print(f"students={students}, name={name}")

ask_for_students("Ahmed","rdff", "dsafsff")
```

```python
def introduce_your_self(**info):  # keyword argument
    print(info)

introduce_your_self(name='noha', work='iti')
introduce_your_self(fullname='ahmed ali')
```
