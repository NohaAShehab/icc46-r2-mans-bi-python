## 🔪 Slice: Get Elements from Index 2 to 4

<div style="background: linear-gradient(135deg, #a8edea 0%, #fed6e3 100%); padding: 15px; border-radius: 8px; margin: 10px 0;">
<p style="margin: 0; color: #2c3e50;"><strong>🔪 Range Slicing:</strong> Use <code style="background: rgba(255,255,255,0.5); padding: 2px 6px; border-radius: 4px;">[start:end]</code> to get a range of elements</p>
</div>

<div style="background: #f3e5f5; padding: 12px; border-left: 4px solid #9c27b0; margin: 10px 0;">
<p style="margin: 0; color: #4a148c;"><strong>📌 Remember:</strong> End index is <strong>exclusive</strong> - <code style="background: #e1bee7; padding: 2px 6px; border-radius: 4px;">[2:5]</code> gets indices 2, 3, and 4</p>
</div>


````python
print(l[2:5])
````

## 🔄 Slice with Step: Get Every 2nd Element

<div style="background: linear-gradient(135deg, #ffecd2 0%, #fcb69f 100%); padding: 15px; border-radius: 8px; margin: 10px 0;">
<p style="margin: 0; color: #2c3e50;"><strong>🔄 Step Slicing:</strong> Use <code style="background: rgba(255,255,255,0.5); padding: 2px 6px; border-radius: 4px;">[start:end:step]</code> to skip elements</p>
</div>

<div style="background: #fff3e0; padding: 12px; border-left: 4px solid #ff9800; margin: 10px 0;">
<p style="margin: 0; color: #e65100;">Example: <code style="background: #ffe0b2; padding: 2px 6px; border-radius: 4px;">[::2]</code> gets every 2nd element (indices 0, 2, 4, 6...)</p>
</div>


````python
## 🔄 Slice with Step: Get Every 2nd Element

```python
print(l[::2])
```
````

## 🔙 Reverse Slice: Get Elements in Reverse Order
