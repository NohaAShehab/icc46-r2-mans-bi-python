## 📖 Reading Excel Files with `openpyxl`

`openpyxl` is a Python library for reading and writing Excel 2010 xlsx/xlsm/xltx/xltm files.

### Steps:
1. Import `openpyxl`
2. Load the workbook using `load_workbook()`
3. Access the active sheet
4. Read cells using `cell(row, column)` method


```python
import openpyxl
path = ".xlsx"

# To open the workbook
# workbook object is created
wb_obj = openpyxl.load_workbook(path)

# Get workbook active sheet object
# from the active attribute
sheet_obj = wb_obj.active

cell_obj = sheet_obj.cell(row=1, column=1)

print(cell_obj.value)
```

### Reading a Specific Cell

Access any cell by specifying its row and column number:
- Row 4, Column 3 = Cell C4


```python
import openpyxl
path = "grade.xlsx"
# To open the workbook
# workbook object is created
wb_obj = openpyxl.load_workbook(path)

# Get workbook active sheet object
# from the active attribute
sheet_obj = wb_obj.active

cell_obj = sheet_obj.cell(row=4, column=3)

print(cell_obj.value)
```

## 📋 Reading Cell Ranges

You can read multiple cells at once using slice notation like `'A1:D12'`:
- This reads all cells from A1 to D12
- Returns a tuple of cell objects for each row


```python
import openpyxl

# Give the location of the file
path = "grade.xlsx"

wb_obj = openpyxl.load_workbook(path)

sheet_obj = wb_obj.active

cell_obj = sheet_obj['A1': 'D12']

for cell1, cell2, cell3, cell4 in cell_obj:
    print(cell1.value, cell2.value, cell3.value, cell4.value)
```

## 🔄 List Comprehensions

List comprehensions provide a concise way to create lists in Python.

### Basic List Comprehension
Creates a list of numbers from 0 to 9.


```python
# list comprehension
l = [num for num in range(10)]
print(l)
```

### List Comprehension with Condition

Filter elements based on a condition (e.g., only even numbers).


```python
l = [num for num in range(10) if num%2==0]
print(l)
```

## 🔄 Converting Excel to List of Dictionaries

This function converts Excel data into a more Python-friendly format:
- Reads headers from the first row
- Converts each data row into a dictionary
- Uses list comprehensions and `zip()` to pair headers with values

**Note**: `openpyxl` reads formulas as strings (e.g., `=1+A2`), not calculated values.


```python
from openpyxl import load_workbook

def excel_to_dict_list(file_path):
    wb = load_workbook(file_path)
    ws = wb.active
    # Get headers from first row
    headers = [cell.value.capitalize() for cell in ws[1]]
    # print(headers)

    data = []
    for row in ws.iter_rows(min_row=2, values_only=True):
        row_dict = {}
        # print(headers, row, zip(headers, row))
        for x, y in zip(headers, row):
            row_dict[x] = y
        data.append(row_dict)

    return data

list_of_dicts = excel_to_dict_list('grade.xlsx')
print(list_of_dicts)
```

## 🐼 Using Pandas for Excel Operations

`pandas` provides a simpler and more powerful way to work with Excel files:
- `pd.read_excel()` automatically handles headers
- Converts Excel data to a DataFrame
- `df.to_dict('records')` converts DataFrame to list of dictionaries
- **Advantage**: Pandas evaluates formulas automatically, giving calculated values instead of formula strings


```python
# using pandas
import pandas as pd

# Read the Excel file
df = pd.read_excel('grade.xlsx')

# Convert to list of dictionaries
list_of_dicts = df.to_dict('records')

print(list_of_dicts)
```

## 📊 Displaying Data with `tabulate`

The `tabulate` library creates beautiful formatted tables from Python data structures.

### Basic Table
Default format without headers.


```python
# display content of list_of_dics into a table

from tabulate import  tabulate
print(tabulate(list_of_dicts))
```

### Table with Headers

Use `headers='keys'` to display dictionary keys as column headers.


```python
print(tabulate(list_of_dicts, headers='keys'))
```

### Custom Table Formats

`tabulate` supports many table formats:
- `'psql'` - PostgreSQL-style tables
- `'double_outline'` - Double-line borders
- `'grid'`, `'fancy_grid'`, `'html'`, and many more!

Try different formats to find the one that suits your needs.


```python
print(tabulate(list_of_dicts, headers='keys', tablefmt='psql'))
```
