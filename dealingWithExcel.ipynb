```python
from openpyxl import load_workbook

def excel_to_dict_list(file_path):
    wb = load_workbook(file_path)
    ws = wb.active
    # Get headers from first row
    headers = [cell.value.capitalize() for cell in ws[1]]
    # print(headers)

    data = []
    for row in ws.iter_rows(min_row=2, values_only=True):
        row_dict = {}
        # print(headers, row, zip(headers, row))
        for x, y in zip(headers, row):
            row_dict[x] = y
        data.append(row_dict)

    return data

list_of_dicts = excel_to_dict_list('grade.xlsx')
print(list_of_dicts)
```

```python
# using pandas
import pandas as pd

# Read the Excel file
df = pd.read_excel('grade.xlsx')

# Convert to list of dictionaries
list_of_dicts = df.to_dict('records')

print(list_of_dicts)
```

```python
# display content of list_of_dics into a table

from tabulate import  tabulate
print(tabulate(list_of_dicts))
```
